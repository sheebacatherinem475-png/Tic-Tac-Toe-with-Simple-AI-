```python

import random

# Create the game board
board = [" "] * 9

def display_board():
    print()
    for i in range(0, 9, 3):
        print(" " + " | ".join(board[i:i+3]))
        if i < 6:
            print("---+---+---")
    print()

def check_winner(player):
    win_positions = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    return any(
        board[a] == board[b] == board[c] == player
        for a, b, c in win_positions
    )

def ai_move():
    # AI first tries to win
    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            if check_winner("O"):
                return
            board[i] = " "

    # AI blocks the player's winning move
    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            if check_winner("X"):
                board[i] = "O"
                return
            board[i] = " "

    # Otherwise, choose a random empty position
    empty = [i for i in range(9) if board[i] == " "]
    if empty:
        board[random.choice(empty)] = "O"

def play_game():
    global board
    board = [" "] * 9

    print("TIC-TAC-TOE WITH SIMPLE AI")
    print("You are X | Computer is O")
    print("Choose positions from 1 to 9")

    while True:
        print("\nBoard positions:")
        print(" 1 | 2 | 3")
        print("---+---+---")
        print(" 4 | 5 | 6")
        print("---+---+---")
        print(" 7 | 8 | 9")

        display_board()

        try:
            move = int(input("Enter your position (1-9): ")) - 1
        except ValueError:
            print("Please enter a number!")
            continue

        if move < 0 or move > 8 or board[move] != " ":
            print("Invalid position! Try again.")
            continue

        board[move] = "X"

        if check_winner("X"):
            display_board()
            print("Congratulations! You win!")
            break

        if " " not in board:
            display_board()
            print("It's a Draw!")
            break

        ai_move()
        print("Computer played:")
        display_board()

        if check_winner("O"):
            print("Computer wins! Try again.")
            break

        if " " not in board:
            print("It's a Draw!")
            break

play_game()
```
